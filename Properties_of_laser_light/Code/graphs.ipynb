```python
import numpy as np 
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
from scipy.signal import butter, filtfilt
```

```python
data = np.loadtxt("../Data/warmup.txt",skiprows=1)

plt.plot(data[:,0],data[:,1],c='royalblue')
plt.xlabel('Time (s)')
plt.ylabel("Intensity (arb. units)")
plt.savefig('../Figures/warmupnopol',bbox_inches = 'tight')
plt.show()
```

```python
data = np.loadtxt("../Data/warmup.txt",skiprows=1)

result = data[:,1]

dt = data[1,0] - data[0,0]
fs = 1.0 / dt
fc = 0.1

b, a = butter(3, fc / (fs / 2), btype='highpass')
result_hp = filtfilt(b, a, result)

fft = np.fft.rfft(result_hp)
fftfreq = np.fft.rfftfreq(len(result_hp), dt)

plt.plot(fftfreq,np.abs(fft),c='royalblue')
plt.xlabel('Frequency (Hz)')
plt.yticks([])

plt.show()
```
